# Adding Custom Analyses

Learn how to create and integrate custom analyses into the Model Live Diagnostics package.

### Prerequisites
Before starting, ensure you have:
* Started a Model Live Diagnostics session.
* Loaded a catalogue and dataset.

For setup guidance, refer to the [Getting Started Tutorial](./getting_started_tutorial.ipynb).

---

## Analysis Interface

The **Analysis** section provides a dropdown menu of predefined analyses. Selecting an analysis dynamically populates the interface with the required input widgets. 

<img src="../_static/assets/Analysis_section.png" alt="Model Live Diagnostics Analysis Section Interface" style="width: 100%; height: auto;">

By the end of this tutorial, you will know how to implement a custom analysis that automatically generates these interactive UI controls based on user selection. As just one example of this flexibility, you can write generalised analyses that apply to any variable, eliminating the need for redundant functions. Ultimately, any analysis that can be computed with `xarray` and plotted via `matplotlib.pyplot` can be integrated into the analysis section.
